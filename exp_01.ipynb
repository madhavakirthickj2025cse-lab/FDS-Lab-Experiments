```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Sales': [120, 150, 135, 180, 210, 240]
}

df = pd.DataFrame(data)

plt.plot(df['Month'], df['Sales'], marker='o')
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.show()
```

```python
import matplotlib.pyplot as plt

cities = ['Chennai', 'Mumbai', 'Delhi', 'Kolkata', 'Bangalore']
population = [70, 125, 190, 45, 85]

plt.bar(cities, population)
plt.title('Population by City')
plt.xlabel('City')
plt.ylabel('Population (in lakhs)')
plt.show()
```

```python
import pandas as pd

# Structured data
structured_data = pd.DataFrame({
    'Student_ID': [101, 102, 103],
    'Student': ['Arun', 'Meena', 'Karthik'],
    'Marks': [85, 92, 78]
})

print("Structured Data:\n", structured_data)

# Unstructured data
unstructured_data = "Meena is a hardworking student who enjoys learning new subjects and participating in school activities."

print("\nUnstructured Data:\n", unstructured_data)

# Semi-structured data
semi_structured_data = {
    'Student_ID': 101,
    'Student': 'Arun',
    'Details': {
        'Class': '10th Grade',
        'Section': 'A',
        'Hobby': 'Cricket'
    }
}

print("\nSemi-structured Data:\n", semi_structured_data)
```

```python

from cryptography.fernet import Fernet

# Generate encryption key
key = Fernet.generate_key()
cipher_suite = Fernet(key)

# Get input from the user
plain_text = input("Enter the text to encrypt: ").encode()

# Encrypt data
cipher_text = cipher_suite.encrypt(plain_text)

# Decrypt data
decrypted_text = cipher_suite.decrypt(cipher_text)

# Display output
print("Original Data:", plain_text)
print("Encrypted Data:", cipher_text)
print("Decrypted Data:", decrypted_text)

```

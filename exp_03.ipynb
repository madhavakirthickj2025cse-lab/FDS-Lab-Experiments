```python
import pandas as pd
import numpy as np

data = {
    "CustomerID": [201, 202, 203, 204, 205, 206, 207, 208, 209, 209, 210],
    "Age_Group": ["18-25", "26-35", "36-45", "46+", "18-25",
                  "26-35", "36-45", "46+", "18-25", "18-25", "26-35"],
    "Rating(1-5)": [4, 5, 3, 6, 2, 4, -1, 5, 3, 3, 4],
    "Restaurant": ["Dominos", "KFC", "Pizza Hut", "Dominos",
                   "McDonalds", "KFC", "Pizza Hut", "Dominos",
                   "KFC", "KFC", "McDonalds"],
    "FoodPreference": ["veg", "Non-Veg", "Vegetarian", "Veg",
                       "non-Veg", "Veg", "Vegetarian", "Non-Veg",
                       "veg", "Veg", "Non-Veg"],
    "Bill": [650, 900, 1200, 1500, 750, 1100, -300,
             1800, 950, 950, 2200],
    "NoOfPax": [2, 3, 4, 5, 2, 4, -2, 6, 3, 3, -1],
    "EstimatedSalary": [30000, 42000, 50000, 70000, 28000,
                        45000, -5000, 85000, 35000, 35000, 60000]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Remove duplicate rows
df.drop_duplicates(inplace=True)
df.reset_index(drop=True, inplace=True)

# Replace invalid values with NaN
df.loc[df["Bill"] < 0, "Bill"] = np.nan
df.loc[df["EstimatedSalary"] < 0, "EstimatedSalary"] = np.nan
df.loc[df["NoOfPax"] < 1, "NoOfPax"] = np.nan

df.loc[
    (df["Rating(1-5)"] < 1) | (df["Rating(1-5)"] > 5),
    "Rating(1-5)"
] = np.nan

# Standardize food preference values
df["FoodPreference"] = df["FoodPreference"].replace(
    ["veg", "Vegetarian"], "Veg"
)

df["FoodPreference"] = df["FoodPreference"].replace(
    ["non-Veg", "non-veg"], "Non-Veg"
)

# Fill missing values
df["Bill"] = df["Bill"].fillna(df["Bill"].mean())

df["EstimatedSalary"] = df["EstimatedSalary"].fillna(
    df["EstimatedSalary"].mean()
)

df["NoOfPax"] = df["NoOfPax"].fillna(
    df["NoOfPax"].median()
)

df["Rating(1-5)"] = df["Rating(1-5)"].fillna(
    df["Rating(1-5)"].median()
)

print("\nCleaned Dataset:")
print(df)

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
```

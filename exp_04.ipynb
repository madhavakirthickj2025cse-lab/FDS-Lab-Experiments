```python
import numpy as np
import pandas as pd

data = {
    "Country": [
        "UK", "Germany", "Japan", "UK", "France",
        "Japan", "Germany", "France", np.nan, "UK"
    ],

    "Age": [
        22, 35, 40, np.nan, 31, 27, 45, 38, 29, 33
    ],

    "Salary": [
        28000, 55000, 67000, 42000, 61000,
        np.nan, 72000, 49000, 53000, 58000
    ],

    "Purchased": [
        "No", "Yes", "No", "Yes", "Yes",
        "No", "Yes", "No", "Yes", "No"
    ]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)

print("Dataset Information:")
df.info()

print("\nMode of Country:")
print(df["Country"].mode())

# Fill missing country using mode
df["Country"] = df["Country"].fillna(
    df["Country"].mode()[0]
)

# Fill missing age using median
df["Age"] = df["Age"].fillna(
    df["Age"].median()
)

# Fill missing salary using rounded mean
df["Salary"] = df["Salary"].fillna(
    round(df["Salary"].mean())
)

print("\nAfter Data Preprocessing:")
display(df)

print("\nMissing Values After Preprocessing:")
print(df.isnull().sum())
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Different input array
array = np.array([
    15, 18, 20, 22, 24, 26, 28, 30,
    32, 34, 36, 38, 40, 42, 85, 95
])

print("Given Array:")
print(array)

print("Mean:", array.mean())

print("25th Percentile:", np.percentile(array, 25))
print("50th Percentile:", np.percentile(array, 50))
print("75th Percentile:", np.percentile(array, 75))
print("100th Percentile:", np.percentile(array, 100))


# Function to detect outliers
def outDetection(array):
    Q1, Q3 = np.percentile(array, [25, 75])
    IQR = Q3 - Q1

    lr = Q1 - (1.5 * IQR)
    ur = Q3 + (1.5 * IQR)

    return lr, ur


lr, ur = outDetection(array)

print("Lower Range:", lr)
print("Upper Range:", ur)

# Identify outliers
outliers = array[(array < lr) | (array > ur)]

print("Outliers:")
print(outliers)

# Remove outliers
new_array = array[(array >= lr) & (array <= ur)]

print("Array after removing outliers:")
print(new_array)

# Histogram before outlier removal
plt.hist(array, bins=8, edgecolor="black")
plt.title("Before Outlier Removal")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()

# Detect outliers again
lr1, ur1 = outDetection(new_array)

print("New Lower Range:", lr1)
print("New Upper Range:", ur1)

# Final outlier removal
final_array = new_array[
    (new_array >= lr1) & (new_array <= ur1)
]

print("Final Array:")
print(final_array)

# Histogram after outlier removal
plt.hist(final_array, bins=8, edgecolor="black")
plt.title("After Outlier Removal")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

```python
import numpy as np
import pandas as pd

data = {
    'Country': ['UK', 'France', 'Germany', 'UK', 'Germany',
                'France', 'UK', 'Germany', 'France', 'UK'],
    'Age': [22, 35, 29, 48, np.nan, 31, 40, 26, 45, 33],
    'Salary': [28000, 52000, 45000, 85000, 67000,
               49000, np.nan, 38000, 92000, 61000],
    'Purchased': ['No', 'Yes', 'No', 'Yes', 'No',
                  'Yes', 'No', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Fill missing country values
df['Country'] = df['Country'].fillna(
    df['Country'].mode()[0]
)

# Separate features and label
features = df.iloc[:, :-1].values
label = df.iloc[:, -1].values

# Handle missing numerical values
from sklearn.impute import SimpleImputer

age = SimpleImputer(strategy='mean')
salary = SimpleImputer(strategy='mean')

features[:, [1]] = age.fit_transform(features[:, [1]])
features[:, [2]] = salary.fit_transform(features[:, [2]])

# One-hot encode country
from sklearn.preprocessing import OneHotEncoder

oh = OneHotEncoder(sparse_output=False)
Country = oh.fit_transform(features[:, [0]])

print("\nEncoded Country:")
print(Country)

# Combine encoded country, age, and salary
final_set = np.concatenate(
    (Country, features[:, [1, 2]]),
    axis=1
)

print("\nFinal Set:")
print(final_set)

# Standard scaling
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
feat_standard_scaler = sc.fit_transform(final_set)

print("\nStandard Scaled Data:")
print(feat_standard_scaler)

# Min-max scaling
from sklearn.preprocessing import MinMaxScaler

mms = MinMaxScaler(feature_range=(0, 1))
feat_minmax_scaler = mms.fit_transform(final_set)

print("\nMin-Max Scaled Data:")
print(feat_minmax_scaler)

print("\nPurchased Labels:")
print(label)
```

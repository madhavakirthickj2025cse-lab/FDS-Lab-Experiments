```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Custom input data with different values
data = {
    'total_bill': [18.5, 25.2, 42.6, 16.7, 48.3,
                   31.4, 27.8, 13.9, 38.6, 34.5],
    'tip': [2.5, 4.0, 7.5, 2.0, 8.0,
            5.0, 4.5, 1.5, 6.5, 5.5],
    'sex': ['Male', 'Female', 'Male', 'Female', 'Male',
            'Female', 'Male', 'Female', 'Male', 'Female'],
    'day': ['Fri', 'Sat', 'Sun', 'Thur', 'Sat',
            'Sun', 'Fri', 'Thur', 'Sat', 'Sun'],
    'time': ['Dinner', 'Lunch', 'Dinner', 'Lunch', 'Dinner',
             'Dinner', 'Lunch', 'Lunch', 'Dinner', 'Dinner'],
    'size': [2, 4, 5, 2, 6, 3, 3, 2, 4, 5]
}

# Create DataFrame
tips = pd.DataFrame(data)

print("First 5 rows:")
print(tips.head())

# Joint plot
sns.jointplot(data=tips, x='tip', y='total_bill', kind='reg')
plt.show()

# Count of Lunch/Dinner
print("\nTime Value Counts:")
print(tips['time'].value_counts())

# Pair plot
sns.pairplot(tips, hue='day')
plt.show()

# Correlation heatmap
sns.heatmap(
    tips.corr(numeric_only=True),
    annot=True,
    cmap='coolwarm'
)
plt.title('Correlation Heatmap')
plt.show()

# Boxplot - Total Bill
sns.boxplot(data=tips, x='total_bill')
plt.title('Total Bill Distribution')
plt.show()

# Boxplot - Tip
sns.boxplot(data=tips, x='tip')
plt.title('Tip Distribution')
plt.show()

# Count plot - Day
sns.countplot(data=tips, x='day')
plt.title('Count by Day')
plt.show()

# Count plot - Sex
sns.countplot(data=tips, x='sex')
plt.title('Count by Sex')
plt.show()

# Pie chart - Sex
tips['sex'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)
plt.title('Sex Distribution')
plt.ylabel('')
plt.show()

# Bar chart - Sex
tips['sex'].value_counts().plot(kind='bar')
plt.title('Count by Sex')
plt.xlabel('Sex')
plt.ylabel('Count')
plt.show()

# Dinner count by day
sns.countplot(
    data=tips[tips['time'] == 'Dinner'],
    x='day'
)
plt.title('Dinner Count by Day')
plt.show()
```

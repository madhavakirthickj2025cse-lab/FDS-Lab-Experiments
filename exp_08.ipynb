```python
import numpy as np
import pandas as pd

# Different input data
data = {
    'YearsExperience': [
        0.8, 1.5, 2.2, 2.9, 3.5,
        4.1, 4.8, 5.4, 6.0, 6.7,
        7.3, 8.0, 8.7, 9.3, 10.0
    ],
    'Salary': [
        28000, 33000, 39000, 45000, 51000,
        57000, 63000, 69000, 75000, 82000,
        88000, 95000, 101000, 108000, 115000
    ]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)

print("\nDataset Information:")
df.info()

# Remove missing values
df.dropna(inplace=True)

print("\nSummary Statistics:")
print(df.describe())

# Separate features and label
features = df[['YearsExperience']].values
label = df[['Salary']].values

# Split dataset into training and testing data
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    features,
    label,
    test_size=0.2,
    random_state=42
)

# Train linear regression model
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x_train, y_train)

# Evaluate model
print("\nTraining Score:", model.score(x_train, y_train))
print("Testing Score:", model.score(x_test, y_test))
print("Coefficient:", model.coef_)
print("Intercept:", model.intercept_)

# Save model
import pickle

with open('SalaryPred.model', 'wb') as file:
    pickle.dump(model, file)

# Load saved model
with open('SalaryPred.model', 'rb') as file:
    model = pickle.load(file)

# Predict salary
yr_of_exp = float(input("Enter Years of Experience: "))

salary = model.predict(np.array([[yr_of_exp]]))

print(
    "Estimated Salary for",
    yr_of_exp,
    "years of experience is",
    round(salary[0][0], 2)
)
```
